Remove very short reviews

```python
import pandas as pd
import re

# Load dataset
df = pd.read_csv("/content/cleaned_reviews_ratings.csv")

# Assuming the review column is named 'Review'
# 1. Convert to lowercase
df['Review'] = df['Review'].str.lower()

# 2. Remove special characters & extra spaces
df['Review'] = df['Review'].apply(lambda x: re.sub(r'[^a-z0-9\s]', '', str(x)))  # keep only letters, numbers, spaces
df['Review'] = df['Review'].str.strip()  # remove leading/trailing spaces
df['Review'] = df['Review'].replace(r'\s+', ' ', regex=True)  # remove multiple spaces

# 3. Remove short reviews (2–4 words)
df = df[df['Review'].apply(lambda x: len(str(x).split()) > 4)]

# 4. Drop blank/NaN reviews
df = df[df['Review'].notna() & (df['Review'].str.strip() != '')]

# Save cleaned dataset
df.to_csv("cleaned_reviews.csv", index=False)

print("✅ Cleaned: lowercased, special chars removed, blanks dropped, short reviews removed")
```

Remove Special Characters

```python
import pandas as pd
import re

# Load dataset
df = pd.read_csv("/content/reviews_ratings_final.csv")

df = df.rename(columns={
    "rating": "Rating",
    "review_content": "Review"
})

# --- 1. Remove missing values ---
df = df.dropna(subset=['Review', 'Rating'])

# --- 2. Remove exact duplicates (same Review + Rating) ---
df = df.drop_duplicates(subset=['Review', 'Rating'])

# --- 3. Remove conflicting reviews (same text with different ratings) ---
conflicts = df.groupby('Review')['Rating'].nunique()
conflict_reviews = conflicts[conflicts > 1].index
df = df[~df['Review'].isin(conflict_reviews)]

# --- 4. Clean Review column (remove special characters & symbols) ---
def clean_text(text):
    if isinstance(text, str):
        # Keep only letters, numbers, and spaces
        text = re.sub(r'[^A-Za-z0-9 ]+', '', text)
        # Remove extra spaces
        text = re.sub(r'\s+', ' ', text).strip()
        return text.lower()
    return ""

df['Review'] = df['Review'].apply(clean_text)

# Save cleaned dataset
df.to_csv("reviews_ratings_cleaned.csv", index=False)

print("✅ Cleaning complete!")
print("Final shape:", df.shape)
print(df['Rating'].value_counts())
```

Creating the Balanced and imbalanced dataset

```python
import pandas as pd

# Load dataset
df = pd.read_csv("/content/balanced_review_dataset_cleaned_filtered.csv")

# Shuffle for randomness
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Desired counts for imbalanced dataset
imbalanced_counts = {
    1: 1200,
    2: 1000,
    3: 2800,
    4: 2000,
    5: 3000
}

# CREATE IMBALANCED DATASET
imbalanced_parts = []
remaining_parts = []

for rating, count in imbalanced_counts.items():
    rating_df = df[df['rating'] == rating]
    imbalanced_sample = rating_df.sample(n=count, random_state=42)

    imbalanced_parts.append(imbalanced_sample)

    # Remove chosen samples so they are not reused
    remaining = rating_df.drop(imbalanced_sample.index)
    remaining_parts.append(remaining)

imbalanced_df = pd.concat(imbalanced_parts).sample(frac=1, random_state=42).reset_index(drop=True)

# CREATE BALANCED DATASET (2000 each)
balanced_parts = []
for rating, remaining_df in zip(imbalanced_counts.keys(), remaining_parts):
    balanced_sample = remaining_df.sample(n=2000, random_state=42)
    balanced_parts.append(balanced_sample)

balanced_df = pd.concat(balanced_parts).sample(frac=1, random_state=42).reset_index(drop=True)

# SAVE OUTPUTS
imbalanced_df.to_csv("imbalanced_review_dataset.csv", index=False)
balanced_df.to_csv("balanced_review_dataset.csv", index=False)

# Check distributions
print("Imbalanced distribution:\n", imbalanced_df['rating'].value_counts())
print("\nBalanced distribution:\n", balanced_df['rating'].value_counts())
```

Stopword removed words

```python
import pandas as pd
import spacy
from nltk.corpus import stopwords
import nltk
import re

# Download stopwords
nltk.download('stopwords')

# Load dataset
df = pd.read_csv("/content/balanced_review_dataset.csv")

# Load spaCy model
nlp = spacy.load("en_core_web_sm", disable=["parser", "ner", "textcat"])

# Define custom stop words (keep negations)
stop_words = set(stopwords.words('english'))
negations = {"no", "not", "nor", "never", "n't"}
stop_words = stop_words - negations

# Set to keep track of removed words
removed_words_set = set()

# Preprocessing function
def preprocess_doc(doc):
    kept_tokens = []
    for token in doc:
        # Keep alphabetic tokens, exclude stopwords
        if token.is_alpha and token.text not in stop_words:
            kept_tokens.append(token.lemma_)
        else:
            # Only keep alphabetic words in removed set
            if token.is_alpha:
                removed_words_set.add(token.text)
    return " ".join(kept_tokens)

# Process reviews
cleaned_reviews = []
for doc in nlp.pipe(df['review'].astype(str).str.lower().values, batch_size=1000):
    cleaned_reviews.append(preprocess_doc(doc))

df['review'] = cleaned_reviews

# Keep only rating and review
if 'rating' in df.columns:
    df = df[['rating', 'review']]

# Save cleaned dataset
df.to_csv("/content/balanced_review_dataset_cleaned.csv", index=False)

# Print only alphabetic words removed
print("✅ Alphabetic words removed during preprocessing:")
print(sorted(list(removed_words_set)))
```

Finetuning the balanced dataset

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Load Balanced Dataset
bal_train_df = pd.read_csv("/content/train_dataset.csv")
bal_test_df = pd.read_csv("/content/test_dataset.csv")

X_train, y_train = bal_train_df["review"], bal_train_df["rating"]
X_test, y_test = bal_test_df["review"], bal_test_df["rating"]

# TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

results = {}

print("Fine-tuning on BALANCED Dataset\n")

# Logistic Regression
best_acc, best_params = 0, None
for c in [0.01, 0.1, 1, 10]:
    model = LogisticRegression(C=c, max_iter=1000)
    model.fit(X_train_tfidf, y_train)
    acc = accuracy_score(y_test, model.predict(X_test_tfidf))
    if acc > best_acc:
        best_acc, best_params = acc, c
results["Logistic Regression"] = (best_acc, f"C={best_params}")
print(f"Logistic Regression {best_acc:.4f} | C={best_params}")

# SVM
best_acc, best_params = 0, None
for kernel in ['linear', 'rbf', 'poly']:
    for c in [0.1, 1, 10]:
        model = SVC(kernel=kernel, C=c)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (kernel, c)
results["SVM"] = (best_acc, f"kernel={best_params[0]}, C={best_params[1]}")
print(f"SVM {best_acc:.4f} | Params={best_params}")

# Decision Tree
best_acc, best_params = 0, None
for depth in [5, 10, 20, None]:
    for crit in ['gini', 'entropy']:
        model = DecisionTreeClassifier(max_depth=depth, criterion=crit)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (depth, crit)
results["Decision Tree"] = (best_acc, f"depth={best_params[0]}, criterion={best_params[1]}")
print(f"Decision Tree {best_acc:.4f} | Params={best_params}")

# Random Forest
best_acc, best_params = 0, None
for n in [50, 100, 200]:
    for depth in [10, 20, None]:
        model = RandomForestClassifier(n_estimators=n, max_depth=depth)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (n, depth)
results["Random Forest"] = (best_acc, f"n_estimators={best_params[0]}, depth={best_params[1]}")
print(f"Random Forest {best_acc:.4f} | Params={best_params}")

# Naive Bayes
best_acc, best_params = 0, None
for alpha in [0.1, 0.5, 1.0]:
    model = MultinomialNB(alpha=alpha)
    model.fit(X_train_tfidf, y_train)
    acc = accuracy_score(y_test, model.predict(X_test_tfidf))
    if acc > best_acc:
        best_acc, best_params = acc, alpha
results["Naive Bayes"] = (best_acc, f"alpha={best_params}")
print(f"Naive Bayes {best_acc:.4f} | alpha={best_params}")

# Summary
print("\n Best Model for BALANCED Dataset:")
best_model = max(results, key=lambda x: results[x][0])
print(f"{best_model}: {results[best_model][0]:.4f} ({results[best_model][1]})")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB

# Load datasets
bal_train_df = pd.read_csv("/content/train_dataset.csv")
bal_test_df  = pd.read_csv("/content/test_dataset.csv")

X_train, y_train = bal_train_df["review"], bal_train_df["rating"]
X_test, y_test = bal_test_df["review"], bal_test_df["rating"]

# TF-IDF Vectorization (adding bigrams)
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1,2))
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# --- Logistic Regression (tune C and solver) ---
print("\nLogistic Regression (tuning C and solver):")
for C in [0.01, 0.1, 1, 10, 50]:
    for solver in ['liblinear', 'saga']:
        try:
            model = LogisticRegression(C=C, solver=solver, max_iter=1000)
            model.fit(X_train_tfidf, y_train)
            y_pred = model.predict(X_test_tfidf)
            acc = accuracy_score(y_test, y_pred)
            print(f"C={C}, solver={solver} → Accuracy: {acc:.4f}")
        except:
            continue

# --- Decision Tree (tune max_depth and min_samples_split) ---
print("\nDecision Tree (tuning max_depth and min_samples_split):")
for depth in [5, 10, 20, None]:
    for min_split in [2, 5, 10]:
        model = DecisionTreeClassifier(max_depth=depth, min_samples_split=min_split, random_state=42)
        model.fit(X_train_tfidf, y_train)
        y_pred = model.predict(X_test_tfidf)
        acc = accuracy_score(y_test, y_pred)
        print(f"max_depth={depth}, min_samples_split={min_split} → Accuracy: {acc:.4f}")

# --- Random Forest (tune n_estimators and max_depth) ---
print("\nRandom Forest (tuning n_estimators and max_depth):")
for n in [50, 100, 200]:
    for depth in [None, 10, 20]:
        model = RandomForestClassifier(n_estimators=n, max_depth=depth, random_state=42)
        model.fit(X_train_tfidf, y_train)
        y_pred = model.predict(X_test_tfidf)
        acc = accuracy_score(y_test, y_pred)
        print(f"n_estimators={n}, max_depth={depth} → Accuracy: {acc:.4f}")

# --- SVM (tune kernel, C, gamma) ---
print("\nSVM (tuning kernel, C, gamma):")
for kernel in ['linear', 'rbf', 'poly', 'sigmoid']:
    for C in [0.1, 1, 10]:
        for gamma in ['scale', 'auto']:
            try:
                model = SVC(kernel=kernel, C=C, gamma=gamma, probability=True)
                model.fit(X_train_tfidf, y_train)
                y_pred = model.predict(X_test_tfidf)
                acc = accuracy_score(y_test, y_pred)
                print(f"kernel={kernel}, C={C}, gamma={gamma} → Accuracy: {acc:.4f}")
            except:
                continue

# --- Multinomial Naïve Bayes (tune alpha) ---
print("\nMultinomial Naïve Bayes (tuning alpha):")
for alpha in [0.01, 0.1, 0.5, 1.0, 2.0]:
    model = MultinomialNB(alpha=alpha)
    model.fit(X_train_tfidf, y_train)
    y_pred = model.predict(X_test_tfidf)
    acc = accuracy_score(y_test, y_pred)
    print(f"alpha={alpha} → Accuracy: {acc:.4f}")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

# Load Imbalanced Dataset
im_train_df = pd.read_csv("/content/train_imbalanced.csv")
im_test_df = pd.read_csv("/content/test_imbalanced.csv")

X_train, y_train = im_train_df["review"], im_train_df["rating"]
X_test, y_test = im_test_df["review"], im_test_df["rating"]

# TF-IDF
vectorizer = TfidfVectorizer(max_features=5000)
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

results = {}

print("Fine-tuning on IMBALANCED Dataset\n")

# Logistic Regression
best_acc, best_params = 0, None
for c in [0.01, 0.1, 1, 10]:
    model = LogisticRegression(C=c, max_iter=1000)
    model.fit(X_train_tfidf, y_train)
    acc = accuracy_score(y_test, model.predict(X_test_tfidf))
    if acc > best_acc:
        best_acc, best_params = acc, c
results["Logistic Regression"] = (best_acc, f"C={best_params}")
print(f"Logistic Regression {best_acc:.4f} | C={best_params}")

# SVM
best_acc, best_params = 0, None
for kernel in ['linear', 'rbf', 'poly']:
    for c in [0.1, 1, 10]:
        model = SVC(kernel=kernel, C=c)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (kernel, c)
results["SVM"] = (best_acc, f"kernel={best_params[0]}, C={best_params[1]}")
print(f"SVM {best_acc:.4f} | Params={best_params}")

# Decision Tree
best_acc, best_params = 0, None
for depth in [5, 10, 20, None]:
    for crit in ['gini', 'entropy']:
        model = DecisionTreeClassifier(max_depth=depth, criterion=crit)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (depth, crit)
results["Decision Tree"] = (best_acc, f"depth={best_params[0]}, criterion={best_params[1]}")
print(f"Decision Tree {best_acc:.4f} | Params={best_params}")

# Random Forest
best_acc, best_params = 0, None
for n in [50, 100, 200]:
    for depth in [10, 20, None]:
        model = RandomForestClassifier(n_estimators=n, max_depth=depth)
        model.fit(X_train_tfidf, y_train)
        acc = accuracy_score(y_test, model.predict(X_test_tfidf))
        if acc > best_acc:
            best_acc, best_params = acc, (n, depth)
results["Random Forest"] = (best_acc, f"n_estimators={best_params[0]}, depth={best_params[1]}")
print(f"Random Forest {best_acc:.4f} | Params={best_params}")

# Naive Bayes
best_acc, best_params = 0, None
for alpha in [0.1, 0.5, 1.0]:
    model = MultinomialNB(alpha=alpha)
    model.fit(X_train_tfidf, y_train)
    acc = accuracy_score(y_test, model.predict(X_test_tfidf))
    if acc > best_acc:
        best_acc, best_params = acc, alpha
results["Naive Bayes"] = (best_acc, f"alpha={best_params}")
print(f"Naive Bayes {best_acc:.4f} | alpha={best_params}")

# Summary
print("\n Best Model for IMBALANCED Dataset:")
best_model = max(results, key=lambda x: results[x][0])
print(f"{best_model}: {results[best_model][0]:.4f} ({results[best_model][1]})")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB

# Load datasets
im_train_df = pd.read_csv("/content/train_imbalanced.csv")
im_test_df = pd.read_csv("/content/test_imbalanced.csv")

X_train, y_train = im_train_df["review"], im_train_df["rating"]
X_test, y_test = im_test_df["review"], im_test_df["rating"]

# TF-IDF Vectorization
vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1,2))  # adding bigrams
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# --- Logistic Regression (tune C and solver) ---
print("\nLogistic Regression (tuning C and solver):")
for C in [0.01, 0.1, 1, 10, 50]:
    for solver in ['liblinear', 'saga']:
        try:
            model = LogisticRegression(C=C, solver=solver, max_iter=1000)
            model.fit(X_train_tfidf, y_train)
            y_pred = model.predict(X_test_tfidf)
            acc = accuracy_score(y_test, y_pred)
            print(f"C={C}, solver={solver} → Accuracy: {acc:.4f}")
        except:
            continue

# --- Decision Tree (tune max_depth, min_samples_split) ---
print("\nDecision Tree (tuning max_depth and min_samples_split):")
for depth in [5, 10, 20, None]:
    for min_split in [2, 5, 10]:
        model = DecisionTreeClassifier(max_depth=depth, min_samples_split=min_split, random_state=42)
        model.fit(X_train_tfidf, y_train)
        y_pred = model.predict(X_test_tfidf)
        acc = accuracy_score(y_test, y_pred)
        print(f"max_depth={depth}, min_samples_split={min_split} → Accuracy: {acc:.4f}")

# --- Random Forest (tune n_estimators, max_depth) ---
print("\nRandom Forest (tuning n_estimators and max_depth):")
for n in [50, 100, 200]:
    for depth in [None, 10, 20]:
        model = RandomForestClassifier(n_estimators=n, max_depth=depth, random_state=42)
        model.fit(X_train_tfidf, y_train)
        y_pred = model.predict(X_test_tfidf)
        acc = accuracy_score(y_test, y_pred)
        print(f"n_estimators={n}, max_depth={depth} → Accuracy: {acc:.4f}")

# --- SVM (tune kernel, C, gamma) ---
print("\nSVM (tuning kernel, C, gamma):")
for kernel in ['linear', 'rbf', 'poly', 'sigmoid']:
    for C in [0.1, 1, 10]:
        for gamma in ['scale', 'auto']:
            try:
                model = SVC(kernel=kernel, C=C, gamma=gamma)
                model.fit(X_train_tfidf, y_train)
                y_pred = model.predict(X_test_tfidf)
                acc = accuracy_score(y_test, y_pred)
                print(f"kernel={kernel}, C={C}, gamma={gamma} → Accuracy: {acc:.4f}")
            except:
                continue

# --- Naïve Bayes (tune alpha) ---
print("\nMultinomial Naïve Bayes (tuning alpha):")
for alpha in [0.01, 0.1, 0.5, 1.0, 2.0]:
    model = MultinomialNB(alpha=alpha)
    model.fit(X_train_tfidf, y_train)
    y_pred = model.predict(X_test_tfidf)
    acc = accuracy_score(y_test, y_pred)
    print(f"alpha={alpha} → Accuracy: {acc:.4f}")
```
